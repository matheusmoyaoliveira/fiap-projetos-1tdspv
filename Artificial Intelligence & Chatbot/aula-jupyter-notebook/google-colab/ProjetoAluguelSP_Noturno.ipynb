```python
X = df.drop(columns=['preco_aluguel'])
y = df['preco_aluguel']
```

```python
categorical_cols = X.select_dtypes(include='object').columns
numerical_cols = X.select_dtypes(include=['int', 'float']).columns
```

```python
from sklearn.preprocessing import OrdinalEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
# ENCONDING DAS COLUNAS CATEGORICAS - Label encoder
```

```python
# NORMALIZAÇAO DAS COLUNAS NUMERICAS
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1), categorical_cols),
        ("num", StandardScaler(), numerical_cols)
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from xgboost import XGBRegressor
```

```python
model = XGBRegressor(random_state=42)

pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                           ('regressor', model)])
```

```python
pipeline.fit(X,y)
```

```python
import pickle

# Save the pipeline to a pickle file
with open('model.pkl', 'wb') as f:
    pickle.dump(pipeline, f)

print("Pipeline saved successfully as housing_price_prediction_pipeline.pkl")
```
